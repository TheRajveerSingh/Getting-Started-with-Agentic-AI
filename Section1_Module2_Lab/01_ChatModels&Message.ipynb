```python
# You can also use this section to suppress warnings generated by your code:
def warn(*args, **kwargs):
    pass
import warnings
warnings.warn = warn
warnings.filterwarnings('ignore')
import os
os.environ['ANONYMIZED_TELEMETRY'] = 'False'
#Just turns off background data collection by chromadb

from ibm_watsonx_ai.foundation_models import ModelInference
from ibm_watsonx_ai.metanames import GenTextParamsMetaNames as GenParams
from ibm_watsonx_ai.foundation_models.utils.enums import ModelTypes
from ibm_watson_machine_learning.foundation_models.extensions.langchain import WatsonxLLM
```

```python
model_id = 'ibm/granite-4-h-small' 

parameters = {
    GenParams.MAX_NEW_TOKENS: 256,  # this controls the maximum number of tokens in the generated output
    GenParams.TEMPERATURE: 0.2, # this randomness or creativity of the model's responses 
}

credentials = {
    "url": "https://us-south.ml.cloud.ibm.com"
    # "api_key": "your api key here"
    # uncomment above and fill in the API key when running locally
}

project_id = "skills-network"

model = ModelInference(
    model_id=model_id,
    params=parameters,
    credentials=credentials,
    project_id=project_id
)
```

```python
msg = model.generate("In today's sales meeting, we ")
print(msg['results'][0]['generated_text'])
'''
This is just for my knowledge:
This is using the raw IBM model directly — model.generate() returns a big messy dictionary like this:
{
    'results': [
        {
            'generated_text': 'Dogs are mans best friend',
            'generated_tokens': 6,
            'input_token_count': 10,
            ...
        }
    ]
}
So you're basically saying:

msg['results'] → go to the 'results' key
[0] → take the first result
['generated_text'] → just give me the actual text

It's like receiving a wrapped package and having to unwrap it yourself to get to the actual gift inside. '''
```

## So a LLM can give answer, no prob. But if we want to have the LLM to give answer in our way, by having some 'context' about the type of instruction and problem, we can make the LLM a Chat Model by simply wrapping it. And then using Chat Messages (like human, ai, system) we can invoke the answer generated by the llm, by using our own context.


```python
llm = WatsonxLLM(model = model)
```

```python
print(llm.invoke("Who is man's best friend?"))
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
```

```python
msg = llm.invoke(
    [
        SystemMessage(content="You are a helpful AI bot that assists a user in choosing the perfect book to read in one short sentence"),
        HumanMessage(content="I enjoy mystery novels, what should I read?")
    ]
)
print(type(msg))
print(repr(msg))
```

```python
print(msg)
```

```python
msg = llm.invoke(
    [
        SystemMessage(content="You are a supportive AI bot that suggests fitness activities to a user in one short sentence"),
        HumanMessage(content="I like high-intensity workouts, what should I do?"),
        AIMessage(content="You should try a CrossFit class"),
        HumanMessage(content="How often should I attend?")
    ]
)
```

```python
print(msg)
```

```python
msg = llm.invoke(
    [
        HumanMessage(content="What month follows June?")
    ]
)
```
