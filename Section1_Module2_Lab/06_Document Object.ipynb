## Document Objects

This Part of the lab is about using "Document Objects". <br>
Basically, Document objects are just wrappers/containers for large texts (not necessarily large, but yeah, we use it for very large texts, such as those coming from Docs, or pdfs).<br>
And its necessary, as raw strings just store text as, you know, just "text"..<br>
But when the Document is too large, and texts are too much, then, we need a wrapper, which not only takes the text, but also stores them in an organized and labelled manner.<br>
Like, this is the format for a Document Object:<br>
document_object= =  Document(<br>
    page_content="Python is a programming language...",<br>
    metadata={"source": "python.pdf", "page": 1}<br>
)<br>
.......<br>
As we can see here, it has the 'page_content' and also the 'metadata'. <br>
Using just Plain Strings will just store the big text, but it'll be very hard to understand from where a particular text came from.<br>
On the other hand, since 'Document objects' store texts in an organized way, it divides the huge texts from a document in 'chunks'. So even if we have 100 chunks from a PDF, we can easily know:<br>
Which chunk came from which document?<br>
Which page was it on?<br>
What was the source URL?<br>
Without metadata we'd just have 100 anonymous strings with no way to trace them back.<br>
So when the AI answers a question using chunk 47 from our PDF, it'll be able to say: "This answer came from page 3 of someuploaded-paper.pdf"<br>
That's only possible because the chunk is wrapped in a Document object with metadata tracking where it came from.<br>
#So Document objects serve two purposes: They Sarry the text through the pipeline & Track where it came from via metadata.<br>

## Document Loaders: PDF Loader and URL Loader
What is a Document loader?<br>
Basically, if the doc is in the form of PDF in the internet, we can store it in a document loader using PyPDFLoader. <br>
Similarly, if the doc is in the form of a url or website, we can store it by using WebBaseLoader.<br>
Then, this loader is loaded/saved into the Document object, in the format: <br>
document object = document loader.load() <br>
Now, we can call or print the content in that document accordingly.<br>
For Example;<br>
1. Printing content from page 2 of a document, but in raw form: document[2]<br>
2. Printing the first 1000 words from page 1 of the document loaded, but raw: document[1].page_content[:1000]<br>
3. Printing the first 1000 words from page 1 of the document loaded, but in the proper format, just like in the document: print(document[1].page_content[:1000])<br>

## PDF Loader

```python
# Import the PyPDFLoader class from langchain_community's document_loaders module
# This loader is specifically designed to load and parse PDF files
from langchain_community.document_loaders import PyPDFLoader

# We create a PyPDFLoader instance by passing the URL of the PDF file
# The loader will download the PDF from the specified URL and prepare it for loading
loader = PyPDFLoader("https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/96-FDF8f7coh0ooim7NyEQ/langchain-paper.pdf")

# Call the load() method to:
# 1. Download the PDF if needed
# 2. Extract text from each page
# 3. Create a list of Document objects, one for each page of the PDF
# Each Document will contain the text content of a page and metadata including page number
document = loader.load()
```

## URL Loader

```python
# Import the WebBaseLoader class from langchain_community's document_loaders module
# This loader is designed to scrape and extract text content from web pages
from langchain_community.document_loaders import WebBaseLoader

# Create a WebBaseLoader instance by passing the URL of the web page to load
# This URL points to the LangChain documentation's introduction page
loader = WebBaseLoader("https://python.langchain.com/v0.2/docs/introduction/")

# Call the load() method to:
# 1. Send an HTTP request to the specified URL
# 2. Download the HTML content
# 3. Parse the HTML to extract meaningful text
# 4. Create a list of Document objects containing the extracted content
web_data = loader.load()

# Print the first 1000 characters of the page content from the first Document
# This provides a preview of the successfully loaded web content
# web_data[0] accesses the first Document in the list
# .page_content accesses the text content of that Document
# [:1000] slices the string to get only the first 1000 characters
print(web_data[0].page_content[:1000])
```

How Do Text Splitters Work?<br>
Text Splitters are crucial for RAG. A PDF might be 50 pages. You can't send all 50 pages to the AI at once because:<br>
1. LLMs have a token limit (context window)<br>
2. Sending everything is slow and expensive<br>
3. The AI gets confused with too much information<br>

So we split it into small chunks:<br>
Example:<br>
Page 1 (2000 words)<br>
    ↓ split<br>
Chunk 1 (200 chars)<br>
Chunk 2 (200 chars)  ← 20 char overlap with chunk 1<br>
Chunk 3 (200 chars)  ← 20 char overlap with chunk 2<br>
The overlap is important as it makes sure context isn't lost at the boundary between chunks.<br>
It happens something like:<br>
Chunk 1: "...LangChain is a framework for building"<br>
Chunk 2: "for building AI applications using large language models..."<br>
.......<br>
There are 3 types of Text Splitters: <br>
1. CharacterTextSplitter → splits by character count (best for: Clean, well-formatted text with consistent newlines) <br>
2. RecursiveCharacterTextSplitter → tries paragraphs first, then sentences, then characters (best for: Most real-world use cases — messy web pages, mixed formatting. Most commonly used in RAG) <br>
3. TokenTextSplitter → splits by token count (best for: When you need precise token counts — important when working close to model context limits) <br>

## Text Splitter

## 1. CharacterTextSplitter
CharacterTextSplitter is the simplest method of splitting the content. These splits are based on characters (by default "\n\n") and measures chunk length by number of characters.

```python
# Import the CharacterTextSplitter class from langchain.text_splitter module
# Text splitters are used to divide large texts into smaller, manageable chunks
from langchain.text_splitter import CharacterTextSplitter

# Create a CharacterTextSplitter with specific configuration:
# - chunk_size=200: Each chunk will contain approximately 200 characters
# - chunk_overlap=20: Consecutive chunks will overlap by 20 characters to maintain context
# - separator="\n": Text will be split at newline characters when possible
text_splitter = CharacterTextSplitter(chunk_size=200, chunk_overlap=20, separator="\n")

# Split the previously loaded document (PDF or other text) into chunks
# The split_documents method:
# 1. Takes a list of Document objects
# 2. Splits each document's content based on the configured parameters
# 3. Returns a new list of Document objects where each contains a chunk of text
# 4. Preserves the original metadata for each chunk
chunk1 = text_splitter.split_documents(document)
# Print the total number of chunks created
# This shows how many smaller Document objects were generated from the original document(s)
# The number depends on the original document length and the chunk_size setting
print(len(chunk1))
```

## NOTES:
I could understand how the 3 types of text splitters work, so here is what I got from the internet:<br>
Text splitters break massive documents into smaller, manageable chunks so Large Language Models (LLMs) can process them without exceeding their context limits and generate higher-quality outputs.<br>
Here is how each of the three splitters processes text: <br>
1. CharacterTextSplitter → Splits by character count<br>
This is the simplest and most rigid method. It divides text mechanically based on a hardcoded number of characters (like a string length in programming).<br>
How it works: If you set your chunk size to (100) characters, it counts exactly (100) characters and slices the text, regardless of whether it is in the middle of a sentence or a word.<br>
Best used for: Highly structured, plain text where you need absolute, consistent length control.<br>
2. RecursiveCharacterTextSplitter → Tries paragraphs, sentences, then characters<br>
This is the default and recommended choice for most use cases. Instead of bluntly cutting off text, it looks at the natural structure of your document.<br>
How it works: It tries to split using large boundaries first (like double newlines, which indicate paragraphs). If the resulting paragraph chunk is still too big, it moves to the next smaller boundary (single newlines or sentences), then words, and finally individual characters.<br>
Best used for: Standard documents, books, and articles. It keeps whole thoughts, paragraphs, and sentences intact as much as possible.<br>
3. TokenTextSplitter → Splits by token count<br>
LLMs don't read words or characters; they read "tokens" (chunks of words or sub-words, where (1) token  is roughly 4 characters or (0.75) words). This method splits text exactly by how LLMs understand it.<br>
How it works: It breaks text according to an LLM tokenizer. If your limit is (1000) tokens, it ensures each chunk is strictly under (1000) tokens, preventing truncation errors in the AI model.<br>
Best used for: Environments where you are dealing with strict model input/context limits and need flawless compatibility with tools like OpenAI's models or Claude.<br>

# Trying All Three types of Text Splitters:

```python
from langchain.text_splitter import CharacterTextSplitter, RecursiveCharacterTextSplitter, TokenTextSplitter

sample_text = [Document(page_content="""LangChain is a framework for building LLM applications.
It provides tools for prompt templates, chains, and agents.
You can load PDFs, websites, and other documents.
Then split them into chunks for RAG applications.
RAG stands for Retrieval Augmented Generation.""")]

# Splitter 1 - CharacterTextSplitter
# Splits ONLY at the separator character (\n here)
# If no \n found, keeps chunk as-is even if too big
char_splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=10,
    separator="\n"
)
char_chunks = char_splitter.split_documents(sample_text)
print("=== CharacterTextSplitter ===")
print(f"Number of chunks: {len(char_chunks)}")
for i, chunk in enumerate(char_chunks):
    print(f"Chunk {i}: '{chunk.page_content}'")
    print(f"Length: {len(chunk.page_content)}\n")
```

```python
# Splitter 2 - RecursiveCharacterTextSplitter
# Tries multiple separators in order: \n\n, \n, " ", ""
# Much smarter - keeps trying smaller splits until size is met
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=10
)
recursive_chunks = recursive_splitter.split_documents(sample_text)
print("=== RecursiveCharacterTextSplitter ===")
print(f"Number of chunks: {len(recursive_chunks)}")
for i, chunk in enumerate(recursive_chunks):
    print(f"Chunk {i}: '{chunk.page_content}'")
    print(f"Length: {len(chunk.page_content)}\n")
```

```python
# Splitter 3 - TokenTextSplitter
# Splits by tokens (how AI models actually count text)
# 1 token ≈ 4 characters roughly
from langchain.text_splitter import TokenTextSplitter
token_splitter = TokenTextSplitter(
    chunk_size=20,  # 20 tokens ≈ ~80 characters
    chunk_overlap=5
)
token_chunks = token_splitter.split_documents(sample_text)
print("=== TokenTextSplitter ===")
print(f"Number of chunks: {len(token_chunks)}")
for i, chunk in enumerate(token_chunks):
    print(f"Chunk {i}: '{chunk.page_content}'")
    print(f"Length: {len(chunk.page_content)}\n")
```

## Concept for Text Splitters:

If we imagine this is our text:<br>
ABCDEFGHIJKLMNOPQRSTUVWXYZ<br>
With chunk_size=10 and chunk_overlap=3:<br>
Chunk 1: ABCDEFGHIJ<br>
Chunk 2: HIJKLMNOPQ      ← starts 3 back (HI J is overlap)<br>
Chunk 3: OPQRSTUVWX      ← starts 3 back (OPQ is overlap)<br>
Chunk 4: VWXYZ<br>
<br>
Now with real text:<br>
Our text:<br>
LangChain is a framework.<br>
It provides tools.<br>
You can load PDFs.<br>
RAG stands for retrieval.<br>
With chunk_size=50 and separator="\n":<br>
The splitter reads line by line and asks:<br>
<br>
"Can I add this next line without exceeding 50 characters?"<br>
<br>
"LangChain is a framework." → 25 chars, so we can add it<br>
"It provides tools." → 25+19 = 44 chars, so we can add it<br>
"You can load PDFs." → 44+18 = 62 chars, which is too big than '50'!<br>
So Chunk 1 is sealed:<br>
Chunk 1: "LangChain is a framework.<br>
It provides tools."<br>
Then it backs up by chunk_overlap characters and starts Chunk 2 from there.<br>
......<br>
So Chunks:<br>
Chunk 1: LangChain is a framework. It provides<br>
Chunk 2: provides tools. You can load<br>
Chunk 3: load PDFs. RAG stands<br>

```python
from langchain_core.documents import Document
from langchain_community.document_loaders import PyPDFLoader, WebBaseLoader
from langchain.text_splitter import CharacterTextSplitter, RecursiveCharacterTextSplitter

# To load the LangChain paper
paper_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/96-FDF8f7coh0ooim7NyEQ/langchain-paper.pdf"
pdf_loader = PyPDFLoader(paper_url)
pdf_document = pdf_loader.load()

# Load content from LangChain website
web_url = "https://python.langchain.com/v0.2/docs/introduction/"
web_loader = WebBaseLoader(web_url)
web_document = web_loader.load()

# Create two different text splitters
splitter_1 = CharacterTextSplitter(chunk_size=300, chunk_overlap=30, separator="\n")
splitter_2 = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50, separators=["\n\n", "\n", ". ", " ", ""])
#This is something new, we'll talk about this down

# Apply both splitters to the PDF document
chunks_1 = splitter_1.split_documents(pdf_document)
chunks_2 = splitter_2.split_documents(pdf_document)

# Define a function to display document statistics
def display_document_stats(docs, name):
    """Display statistics about a list of document chunks"""
    total_chunks = len(docs)
    total_chars = sum(len(doc.page_content) for doc in docs)
    avg_chunk_size = total_chars / total_chunks if total_chunks > 0 else 0
    
    # Count unique metadata keys across all documents
    all_metadata_keys = set()
    for doc in docs:
        all_metadata_keys.update(doc.metadata.keys())
    
    # Print the statistics
    print(f"\n=== {name} Statistics ===")
    print(f"Total number of chunks: {total_chunks}")
    print(f"Average chunk size: {avg_chunk_size:.2f} characters")
    print(f"Metadata keys preserved: {', '.join(all_metadata_keys)}")
    
    if docs:
        print("\nExample chunk:")
        example_doc = docs[min(5, total_chunks-1)]  # Get the 5th chunk or the last one if fewer
        print(f"Content (first 150 chars): {example_doc.page_content[:150]}...")
        print(f"Metadata: {example_doc.metadata}")
        
        # Calculate length distribution
        lengths = [len(doc.page_content) for doc in docs]
        min_len = min(lengths)
        max_len = max(lengths)
        print(f"Min chunk size: {min_len} characters")
        print(f"Max chunk size: {max_len} characters")

# Display stats for both chunk sets
display_document_stats(chunks_1, "Splitter 1")
display_document_stats(chunks_2, "Splitter 2")
```

# Something to learn:
Check out the line: <br>
splitter_2 = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50, separators=["\n\n", "\n", ". ", " ", ""])<br>
...This was something new..<br>
This is a priority list of where to cut. The splitter tries them in order:<br>
First try:  "\n\n"  (double newline = paragraph break)<br>
If chunk still too big...<br>
Second try: "\n"    (single newline = line break)<br>
If chunk still too big...<br>
Third try:  ". "    (period + space = sentence end)<br>
If chunk still too big...<br>
Fourth try: " "     (single space = word boundary)<br>
If chunk still too big...<br>
Last resort: ""     (cut anywhere, even mid-word)<br>
...<br>
Why this order makes sense:<br>
It tries to keep the most meaningful units together first:<br>
Paragraphs together (best)<br>
Lines together (good)<br>
Sentences together (okay)<br>
Words together (last resort)<br>
Cut anywhere (desperate)<br>
...<br>
Its something like:<br>
First try cutting between paragraphs<br>
If still too big, cut between sentences<br>
If still too big, cut between words<br>
If still too big, cut mid-word<br>
Always trying to preserve meaning as much as possible.<br>
..<br>
That's why RecursiveCharacterTextSplitter is smarter than CharacterTextSplitter — it has multiple fallback strategies instead of just one.<br>
