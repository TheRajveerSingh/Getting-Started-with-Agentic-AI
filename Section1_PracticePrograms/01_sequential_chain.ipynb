```python
from langchain.chains import LLMChain, SequentialChain
from langchain_core.prompts import PromptTemplate
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os

load_dotenv()
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.5,
    api_key=os.getenv("GROQ_API_KEY")
)
```

```python
# Chain 1 - Location to Meal
template1 = """Your job is to come up with a classic dish from the area that the user suggests.
{location}
YOUR RESPONSE:
"""
prompt1 = PromptTemplate(template=template1, input_variables=['location'])
location_chain = LLMChain(llm=llm, prompt=prompt1, output_key='meal')
```

```python
# Chain 2 - Meal to Recipe
template2 = """Given a meal {meal}, give a short and simple recipe on how to make that dish at home.
YOUR RESPONSE:
"""
prompt2 = PromptTemplate(template=template2, input_variables=['meal'])
dish_chain = LLMChain(llm=llm, prompt=prompt2, output_key='recipe')
```

```python
# Chain 3 - Recipe to Time
template3 = """Given the recipe {recipe}, estimate how much time I need to cook it.
YOUR RESPONSE:
"""
prompt3 = PromptTemplate(template=template3, input_variables=['recipe'])
recipe_chain = LLMChain(llm=llm, prompt=prompt3, output_key='time')
```

```python
# Overall Sequential Chain
overall_chain = SequentialChain(
    chains=[location_chain, dish_chain, recipe_chain],
    input_variables=['location'],
    output_variables=['meal', 'recipe', 'time'],
    verbose=True
)

overall_chain.invoke(input={'location': 'China'})
```
